```python
# Bài 1 - Điều kiện bay drone

MAX_WIND_SPEED = 30

wind_speed = float(input('Enter wind speed (km/h): '))
raining_input = input('Is it raining? (y/n): ')
is_raining = raining_input == 'y'

if wind_speed < 0 or (raining_input != 'y' and raining_input !='n'):
    print('Dữ liệu không hợp lệ')

else:
    if wind_speed <= MAX_WIND_SPEED and not is_raining:
        print('Có thể bay')
    else:
        print('Không thể bay')
```

```python
# Bài 2 - Ứng dụng lịch: kiểm tra năm nhuận

year = int(input('Enter year: '))

if year <= 0:
    print('Năm không hợp lệ')

else:
    if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
        print(f'{year}: Năm nhuận - tháng 2 có 29 ngày')
    else:
        print(f'{year}: Không phải năm nhuận - tháng 2 có 28 ngày')
```

```python
# Bài 3 - Xếp loại học lực

gpa = float(input('Nhập điểm trung bình (thang 10): '))

if gpa < 0 or gpa > 10:
    print('Điểm không hợp lệ')

elif gpa >= 8.5:
    print('Xếp loại: Giỏi')

elif gpa >= 7.0:
    print('Xếp loại: Khá')

elif gpa >= 5.0:
    print('Xếp loại; Trung bình')
    
else:
    print('Xếp loại: Yếu')
```

```python
# Bài 4 - Phí gửi xe chung cư

parking_hours= float(input('Nhập số giờ gửi xe: '))

if parking_hours <= 0:
    print('Dữ liệu không hợp lệ')
    exit()

elif parking_hours > 5:
    total_fee = int(parking_hours * 20000)

elif parking_hours >= 2:
    total_fee = int(parking_hours * 10000)

else:
    total_fee = int(parking_hours * 5000)

print(f'Gửi {parking_hours} giở - Phí gửi xe: {total_fee:,} VND')
```

```python
# Bài 5 - Ứng dụng sức khỏe: phân loại BMI

weight = float(input('Nhập cân nặng (kg): '))
heigth = float(input('Nhập chiều cao (m): '))

bmi = weight / heigth ** 2

if weight <= 0 or heigth <= 0:
    print('Dữ liệu không hợp lệ')

else: 
    if bmi >= 30:
        category = 'Béo phì'

    elif bmi >= 25:
        category = 'Thừa cân'

    elif bmi >= 18.5:
        category = 'Bình thường'

    else:
        category = 'Thiếu cân'

    print(f'BMI = {bmi:.2f} => {category}')
```

```python
# Bài 6 - Tính tiền điện bậc thang

kWh = int(input('Nhập số kWh tiêu thụ: '))
print(f'Số điện: {kWh} kWh')

if kWh < 0:
    print('Dữ liệu không hợp lệ')
    exit()

elif kWh <= 50:
    total_fee = kWh * 1700

elif kWh > 50 and kWh <= 100:
    total_fee = 50 * 1700 + (kWh - 50) * 2000

else:
    total_fee = 50 * 1700 + 50 * 2000 + (kWh - 100) * 2500

print(f'Tiền điện phải trả: {total_fee:,} VND')
```

```python
# Bài 7 - Gợi ý món ăn theo ngân sách

budget = int(input('Nhập ngân sách: '))
mode = input('Nhập chế độ ăn (veg/normal): ')

if budget < 0 or (mode != 'veg' and mode != 'normal'):
    print('Dữ liệu không hợp lệ')

elif mode == 'veg' and budget >= 50:
    print('Gợi ý: Cơm chay thập cẩm')

elif mode == 'veg' and budget < 50:
    print('Gợi ý: Bánh mì rau')

elif mode == 'normal' and budget >= 80000:
    print('Gợi ý: Cơm gà')

elif mode == 'normal'and budget >= 30000:
    print('Gợi ý: Bánh mì thịt')

else:
    print('Không đủ tiền, nhịn đói!')
```

```python
# Bài 8 - Kiểm tra tam giác

a = int(input('Nhập cạnh a: '))
b = int(input('Nhập cạnh b: '))
c = int(input('Nhập cạnh c: '))

if a <= 0 or b <= 0 or c <= 0 or a + b <= c or a + c <= b or b + c <= a:
    print('Không phải tam giác')

elif a == b == c:
    print('Tam giác đều')

elif a == b or a == c or b == c:
    print('Tam giác cân')

elif a ** 2 == b ** 2 + c ** 2 or b ** 2 == a ** 2 + c ** 2 or c ** 2 == a ** 2 + b ** 2:
    print('Tam giác vuông')
    
else:
    print('Tam giác thường')

# Cần kiểm tra "tam giác đều" trước "tam giác cân" vì nếu bộ dữ liệu là tam giác đều thì bộ dữ liệu này sẽ thỏa mãn điều kiện của tam giác cân, chương trình sẽ dừng lại và print kết quả là "tam giác cân".
# Nếu đảo thứ tưj thì bộ (4,4,4) cho kết quả là "tam giác cân".
```

```python
# Bài 9 - Xét học bổng

gpa = float(input('Nhập GPA (thang 4.0): '))
ren_luyen = int(input('Nhập điểm rèn luyện (0-100): '))
ky_luat = input('Đã từng bị kỷ luuạt (y/n): ')

has_discipline = ky_luat == 'y'

if gpa < 0 or gpa > 4 or ren_luyen < 0 or ren_luyen > 100:
    print('Dữ liệu không hợp lệ')

else: 
    if has_discipline:
        print ('Không đủ điều kiện xét học bổng')

    else:
        if gpa >= 3.6 and ren_luyen >= 90:
            print('Học bổng toàn phần')

        elif gpa >= 3.2 and ren_luyen >= 80:
            print('Học bổng 50%')

        elif gpa >= 2.8 or ren_luyen >= 90:
            print('Khuyến khích học tập')

        else:
            print('Không đạt')
```

```python
# Bài 11 - Hành lý ký gửi khi đặt vé máy bay

ticket_class = input('Hạng vé (economy/business): ')
baggage_weight = float(input('Số kg hành lý ký gửi: '))
has_membership = input('Có thẻ thành viên (y/n): ')
extra_fee = 0

if baggage_weight < 0 or (has_membership != 'y' and has_membership != 'n'):
    print('Dữ liệu không hợp lệ')

else: 
    if ticket_class == 'economy':
        if baggage_weight <= 20:
            extra_fee = 0
        else:
            if has_membership == 'n':
                extra_fee = int((baggage_weight - 20) * 50000)
            else:
                extra_fee = int((baggage_weight - 20) * 50000 / 2)
        
        print(f'Phụ phi hành lý: {(extra_fee):,} VND')

    elif ticket_class == 'business':
        if baggage_weight <= 40:
            extra_fee = 0
        else:
            extra_fee = int((baggage_weight - 40) * 50000)

        print(f'Phụ phi hành lý: {extra_fee:,} VND')

    else:
        print('Hạng vé không hợp lệ')
```
